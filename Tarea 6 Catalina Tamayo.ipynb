# Tarea 6.8

Use una función interpolante para encontrar una buena aproximación a una gaussiana con $\sigma = 1$ y $\mu = 0$ entre $x = 0$ y $x = 2$. ¿Se puede usar ese mismo polinomio para aproximar la gaussiana en $x = 3$? De una respuesta cuantitativa y argumente.

La función que quiero aproximar es

$$f(x)=\frac{1}{\sqrt{2\pi}}\, e^2{-x^2/2}.$$

Para ello utilizo un polinomio de Lagrange con 7 puntos equidistantes en $[0,2]$, es decir, $x_i=i/3$ con $i=0, \ldots, 6$:

$$P(x) = \sum_{i=0}^6 f(x_i)\, L_{6,i}(x), \qquad L_{6,i}(x) = \prod_{j=0,\,j \neq i}^{6} \frac{x - x_j}{x_i - x_j}.$$

El error de esta interpolación es 

$$f(x) - P(x) = \frac{f^{(7)}(\xi)}{7!}\,\omega(x), \qquad \omega(x) = (x - x_0) (x - x_1)\cdots (x - x_6).$$



#### Verificación numérica

```python
import numpy as np

def gaussiana(x):
    return np.exp(-x**2 / 2) / np.sqrt(2 * np.pi)

def lagrange(x, puntos):
    Lis = np.ones(len(puntos))
    fs = np.ones(len(puntos))
    for i in range (len(puntos)):
        fs[i]=puntos[i][1]
        for j in range(len(puntos)):
            if j != i:
                Lis[i] *= (x - puntos[j][0]) / (puntos[i][0] - puntos[j][0])
    return sum(fs * Lis)

def omega (x, puntos):
    return np.prod([x - p[0] for p in puntos])

puntos = [(xi, gaussiana(xi)) for xi in np.linspace(0,2,7)]

malla= np.linspace(0,2,2001)
error= np.array([abs(lagrange(xi, puntos) - gaussiana(xi)) for xi in malla])
omega_max = max(abs(omega(xi, puntos)) for xi in malla) 

f3 = gaussiana(3.0)
P3 = lagrange (3.0, puntos) 

print(f"Error absoluto máximo en [0,2]: {error.max(): .3e}")
print(f"Error relativo máximo en [0,2]: {(error / gaussiana(malla)).max(): .3e}")
print(f"f(3) = {f3: .5f}")
print(f"P(3) = {P3: .5f}") 
print(f"Error absoluto en x = 3: {abs(P3 - f3): .5f}")
print(f"max|omega| en [0,2] = {omega_max: .4f}")
print(f"omega(3) = {omega(3.0, puntos): .2f}")
```

### Conclusión

El polinomio es una buena aproximación de la gaussiana en $[0,2]$: su error absoluto máximo es de aproximadamente $6.4 \times 10^{-5}$ y su error relativo máximo es de aproximadamente $0.02\%$.

Ese mismo polinomio no se puede usar en $x = 3$ porque entrega $P(3) \approx 0.0160$, mientras que el valor exacto es $f(3) \approx 0.0044$, lo que corresponde a un error absoluto de aproximadamente $0.0115$ y a un error relativo de aproximadamente $260\%$.

La razón es que $x = 3$ está fuera del intervalo de interpolación. Dentro de $[0,2]$, cada punto está cerca de algún $x_i$ y $|\omega(x)|$ no supera $0.0438$. En cambio, en $x = 3$ todos los factores $(x - x_i)$ son grandes y $\omega(3) \approx 82.96$, unas $1900$ veces más.
