# Tarea 3.8

Un sensor recibe en promedio 50 partículas por minuto. Ese sensor está diseñado tal que puede detectar si fue impactado por al menos una partícula, pero no tiene ninguna información sobre cuántas partículas lo impactaron.

El sensor se encendió durante un segundo e indicó impacto. Grafique la distribución de probabilidad de que ese pixel haya sido impactado por $n$ partículas.

El promedio es

$$ 50\ \frac{\text{partículas}}{\text{minuto}}.$$

Para un segundo:

$$\lambda=50\frac{1}{60}= \frac{5}{6}\approx0.8333.$$

Por tanto,

$$ N\sim\operatorname{Poisson}\left(\frac56\right).$$

Sabemos que ocurrió un impacto, es decir,

$$ N\ge1.$$

Entonces no queremos simplemente $P(N=n)$, sino la probabilidad condicional

$$P(N=n|N\ge1).$$

Para $n\ge1$,

$$P(N=n|N\ge1)=\frac{P(N=n)}{P(N\ge1)}.$$

Además,

$$P(N\ge1)=1-P(N=0)=1-e^{-\lambda}.$$

Por tanto,

$$\boxed{P(N=n|N\ge1)= \frac{e^{-\lambda}\lambda^n/n!}{1-e^{-\lambda}},\qquad n=1,2,3,\ldots}$$

Esta es una Poisson condicionada a que el valor cero no ocurrió.

## Gráfico de la distribución de probabilidad 

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import factorial

# Función de probabilidad de Poisson
def poisson_prob(n, lam):
    return (lam**n * np.exp(-lam)) / factorial(n)

# Promedio de partículas por segundo
lam = 50/60

# Probabilidad de que haya ocurrido al menos un impacto
p_impacto = 1 - poisson_prob(0, lam)

# Valores de n que se mostrarán
n_vals = np.arange(1, 8)

# Probabilidad condicionada a que N >= 1
p_cond = np.array([
    poisson_prob(n, lam) / p_impacto
    for n in n_vals
])

# Mostrar probabilidades
for n, p in zip(n_vals, p_cond):
    print(f"P(N={n} | N>=1) = {p:.8f}")

print(f"\nSuma mostrada (n=1,...,7) = {p_cond.sum():.8f}")

# Gráfico
plt.figure(figsize=(8, 5))
plt.bar(n_vals, p_cond)

plt.xlabel("Número de partículas n")
plt.ylabel("P(N=n | N≥1)")
plt.title("Número de partículas dado que el sensor indicó impacto")

plt.xticks(n_vals)
plt.show
plt.grid(axis="y", alpha=0.3)
plt.grid()
plt.show()
```

dicionada a que el valor cero no ocurrió.
