# Tarea 4.10

Discuta el teorema central del límite en el caso en el cual las variables no son idénticamente distribuidas. Por ejemplo, si tenemos muchas variables $x_i$, ¿cómo podría fallar el teorema si los cumulantes dependen de $i$?


Sean $x_1, x_2, \ldots, x_N$ variables aleatorias independientes, pero no necesariamente idénticamente distribuidas. Definimos

$$S_N = \sum_{i=1}^{N} x_i$$

con medias $\mu_i$ y varianzas $\sigma_i^2$. Entonces,

$$\mathbb{E}[S_N] = \sum_{i=1}^{N} \mu_i$$

y

$$\mathrm{Var}(S_N) = \sum_{i=1}^{N} \sigma_i^2.$$

Para estudiar si la suma se acerca a una distribución gaussiana, se normaliza:

$$Y_N =\frac{S_N - \sum_{i=1}^{N} \mu_i}{\sqrt{\sum_{i=1}^{N} \sigma_i^2}}.$$

Como las variables son independientes, sus cumulantes se suman. Si $\kappa_{m,i}$ es el cumulante de orden $m$ de $x_i$, entonces para $m \geq 3$:

$$\kappa_m(Y_N)=\frac{\sum_{i=1}^{N} \kappa_{m,i}}{\left(\sum_{i=1}^{N} \sigma_i^2\right)^{m/2}}.$$

Por lo tanto, no es necesario que todas las variables tengan la misma distribución. Lo importante es que los cumulantes de orden superior se vuelvan despreciables al aumentar $N$:

$$\frac{\sum_{i=1}^{N} \kappa_{m,i}}{\left(\sum_{i=1}^{N} \sigma_i^2\right)^{m/2}}\longrightarrow 0,\qquad m \geq 3.$$

Si esto ocurre, la suma normalizada puede tender a una distribución gaussiana. Si alguno de los cumulantes de orden superior no desaparece, el teorema puede fallar.

## Ejemplo de cómo puede fallar

Tomemos

$$x_i = 2^{i-1}\varepsilon_i,$$

donde cada $\varepsilon_i$ vale $+1$ o $-1$ con probabilidad $1/2$.

Para estas variables,

$$\kappa_{2,i} = 4^{i-1},\qquad\kappa_{4,i} = -2\,16^{i-1}.$$

El cuarto cumulante de la suma normalizada es

$$\kappa_4(Y_N)=\frac{-2\sum_{i=1}^{N}16^{i-1}}{\left(\sum_{i=1}^{N}4^{i-1}\right)^2}.$$

Usando las sumas geométricas,

$$\kappa_4(Y_N)=-2\,\frac{(16^N-1)/15}{\left[(4^N-1)/3\right]^2}.$$

Cuando $N \to \infty$,

$$\kappa_4(Y_N) \longrightarrow -\frac{6}{5} \neq 0.$$

Por lo tanto, la suma normalizada no puede tender a una distribución gaussiana, porque una distribución gaussiana tiene todos sus cumulantes de orden mayor que 2 iguales a cero.

La razón es que las variables crecen tan rápido con $i$ que los últimos términos dominan la suma. En ese caso no se produce el efecto de promedio entre muchas contribuciones pequeñas que permite que aparezca el comportamiento gaussiano.

### Verificación numérica del cuarto cumulante normalizado#

```python

def kappa4_normalizado(N):
    suma_var = sum(4**j for j in range(N))
    suma_k4 = -2 * sum(16**j for j in range(N))
    return suma_k4 / suma_var**2

for N in [2, 4, 8, 16]:
    print(f"N = {N:2d}  ->  kappa_4 = {kappa4_normalizado(N):.6f}")

print("Límite esperado:", -6/5)
```

## Conclusión

El teorema central del límite puede seguir siendo válido aunque las variables $x_i$ no sean idénticamente distribuidas. Lo esencial es que, después de normalizar la suma por su desviación estándar, los cumulantes de orden $m \geq 3$ tiendan a cero.

Si los cumulantes dependen de $i$ y crecen lo suficientemente rápido como para que alguno de esos cumulantes normalizados no desaparezca, la suma puede no aproximarse a una gaussiana. En el ejemplo anterior, el cuarto cumulante tiende a $-6/5$ en lugar de cero, por lo que el límite no es normal.
