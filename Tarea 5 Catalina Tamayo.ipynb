# Tarea 5.7
 Demuestre que el número de condicionamiento de una matriz satisface

$$\kappa(AB) \leq \kappa(A)\kappa(B).$$

Encuentre además un ejemplo donde no se cumple la igualdad.

## Definición

El número de condicionamiento de una matriz invertible se define como

$$\kappa(A)=\|A\|\,\|A^{-1}\|.$$

Además, para una norma matricial submultiplicativa se cumple

$$\|AB\|\leq \|A\|\|B\|.$$

## Demostración

Sean $A$ y $B$ matrices invertibles. El número de condicionamiento del producto es

$$\kappa(AB)=\|AB\|\,\|(AB)^{-1}\|.$$

Como

$$(AB)^{-1}=B^{-1}A^{-1},$$

entonces

$$\kappa(AB)=\|AB\|\,\|B^{-1}A^{-1}\|.$$

Usando la propiedad submultiplicativa de la norma,

$$\|AB\|\leq \|A\|\|B\|$$

y

$$\|B^{-1}A^{-1}\|\leq \|B^{-1}\|\|A^{-1}\|.$$

Por lo tanto,

$$\kappa(AB)\leq \|A\|\|B\|\|B^{-1}\|\|A^{-1}\|.$$

Reordenando los factores,

$$\kappa(AB)\leq \kappa(A)\kappa(B)-.$$

Esto demuestra la desigualdad pedida.

## Ejemplo donde no se cumple la igualdad

Tomemos

$$A=\begin{pmatrix}2&0\\0&1\end{pmatrix},\qquad B=\begin{pmatrix}1&0\\0&2\end{pmatrix}.$$

Usando la norma 2,

$$\kappa(A)=2,\qquad \kappa(B)=2,$$

de modo que

$$\kappa(A)\kappa(B)=4.$$

Sin embargo,

$$AB=\begin{pmatrix}2&0\\0&2\end{pmatrix}=2I.$$

Como una matriz proporcional a la identidad tiene número de condición igual a 1,

$$\kappa(AB)=1.$$

Entonces

$$\kappa(AB)=1<4=\kappa(A)\kappa(B)$$

por lo que en este caso la desigualdad es estricta y no se cumple la igualdad.

## Conclusión

Se demostró que

$$\kappa(AB)\leq \kappa(A)\kappa(B$$

y el ejemplo anterior muestra que la iguald.ad no siempre se cumple, ya que en ese caso

$$\kappa(AB)=1\quad\text{y}\quad\kappa(A)\kappa(B)=4.$$
